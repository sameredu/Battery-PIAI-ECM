# 04. Remaining Useful Life (RUL) Prediction
This notebook trains a Random Forest regressor to predict battery Remaining Useful Life (RUL) and evaluates on cell B0018.

```python
import os
import sys
import pandas as pd
sys.path.append('..')

from src.config import COMBINED_PARAMS_CSV, BATTERY_FILES
from src.data_loader import load_full_capacity

df_params = pd.read_csv(COMBINED_PARAMS_CSV)

# Load full capacity datasets for RUL EOL tracking
cap_dfs = []
for key, fpath in BATTERY_FILES.items():
    if os.path.exists(fpath):
        cap_dfs.append(load_full_capacity(fpath, key))
df_full_capacity = pd.concat(cap_dfs, ignore_index=True)
```
